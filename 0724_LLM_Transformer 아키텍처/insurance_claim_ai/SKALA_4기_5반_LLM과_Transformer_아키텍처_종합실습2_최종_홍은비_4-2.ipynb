# 5. Transformer 이해

수업에서 학습한 내용을 프로젝트와 연결하여 설명합니다.

## Tokenization

(작성)

- 입력한 사고 내용과 Prompt는 서브워드 단위의 토큰으로 나뉘고 토큰 ID로 변환됨
- 한국어 문장이 길어지면 토큰 수와 API 비용, 응답시간도 함께 증가함
- 불필요한 토큰을 줄이기 위해 LLM Agent별 Prompt를 역할 중심으로 작성하고 검색 문서 수를 제한함

---

## Embedding

(작성)

- 토큰을 의미가 담긴 숫자 벡터로 변환해 문장 사이의 의미적 유사도를 계산함
- text-embedding-3-small로 판례, 법률, 약관 문서를 벡터로 변환함
- 입력된 사고와 의미가 가까운 문서를 Chroma에서 찾아 LLM에 근거 자료로 전달함

---

## Self-Attention

(작성)

- Self-Attention은 Query, Key, Value를 이용해 문장 안의 토큰 관계를 계산함
- Prompt 앞부분의 역할과 주의사항을 뒤쪽의 사고 내용, 검색 근거와 연결해 해석함
- 입력이 길어질수록 연산량이 늘어나므로 필요한 검색 결과만 LLM에 전달함

---

## Next Token Prediction

(작성)

- LLM은 앞에서 입력되거나 생성된 토큰을 바탕으로 다음 토큰을 하나씩 예측함
- 이 과정을 반복하면서 사고 분석 결과와 최종 리포트를 완성함
- 판례 분석 Agent, 법률 분석 Agent, 보고서 작성 Agent가 LLM을 호출하므로 호출 횟수와 출력 길이가 전체 처리 시간과 비용에 영향을 줌

---

## 프로젝트와 Transformer의 연관성

Transformer의 어떤 특징이 프로젝트에 활용되었는지 설명합니다.

(작성)

- Transformer의 문맥 처리 능력을 이용해 긴 사고 내용과 여러 근거 문서를 함께 분석
- Tokenization → Embedding → Self-Attention → Next Token Prediction 과정이 각 LLM 호출에서 수행됨
- RAG 검색으로 관련 근거를 먼저 찾고, 구조화된 Prompt로 결과 형식을 일정하게 유지함

(작성)

- 유사 판례가 검색되더라도 실제 과실비율과 보험금은 담당자의 추가 확인이 필요함
