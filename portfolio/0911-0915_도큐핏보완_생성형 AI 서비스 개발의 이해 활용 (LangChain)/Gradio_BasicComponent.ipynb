````python
import gradio as gr
import time
from datetime import datetime
````

````python
# ==========================================
# 실습 1: Markdown 컴포넌트
# ==========================================

def markdown_demo():
    """Markdown 컴포넌트 데모"""
    with gr.Blocks(theme=gr.themes.Soft()) as demo:
        gr.Markdown("""
        # Markdown 컴포넌트 실습

        Markdown을 사용하여 **다양한 스타일**의 텍스트를 표현할 수 있습니다.
        """)

        gr.Markdown("---")  # 구분선

        with gr.Tab("기본 문법"):
            gr.Markdown("""
            ## 제목 (Headers)
            # H1 - 가장 큰 제목
            ## H2 - 두 번째 제목
            ### H3 - 세 번째 제목

            ## 텍스트 스타일
            - **굵게**: `**텍스트**`
            - *기울임*: `*텍스트*`
            - ***굵게 + 기울임***: `***텍스트***`
            - ~~취소선~~: `~~텍스트~~`
            - `코드`: \`코드\`

            ## 리스트
            ### 순서 없는 리스트
            - 항목 1
            - 항목 2
              - 하위 항목 2-1
              - 하위 항목 2-2

            ### 순서 있는 리스트
            1. 첫 번째
            2. 두 번째
            3. 세 번째
            """)

        with gr.Tab("고급 기능"):
            gr.Markdown("""
            ## 링크와 이미지
            [Gradio 공식 사이트](https://gradio.app)

            ## 인용구
            > "Gradio는 머신러닝 모델을 위한 최고의 인터페이스 도구입니다."
            > - Gradio 사용자

            ## 코드 블록
            ```python
            import gradio as gr

            def hello(name):
                return f"안녕하세요, {name}님!"

            gr.Interface(fn=hello, inputs="text", outputs="text").launch()
            ```

            ## 표 (Table)
            | 컴포넌트 | 용도 | 난이도 |
            |---------|------|--------|
            | Textbox | 텍스트 입출력 | ⭐ |
            | Button | 액션 트리거 | ⭐ |
            | Tab | 탭 구조 | ⭐⭐ |

            ## 체크박스
            - [x] 완료된 작업
            - [ ] 진행 중인 작업
            - [ ] 예정된 작업
            """)

        with gr.Tab("동적 Markdown"):
            gr.Markdown("### 버튼을 클릭하면 Markdown이 업데이트됩니다")

            update_btn = gr.Button("현재 시간 표시", variant="primary")
            dynamic_md = gr.Markdown("시간이 여기에 표시됩니다.")

            def update_time():
                current_time = datetime.now().strftime("%Y-%m-%d %H:%M:%S")
                return f"""
                ## 현재 시간

                **{current_time}**

                > 버튼을 다시 클릭하면 시간이 업데이트됩니다.
                """

            update_btn.click(fn=update_time, outputs=dynamic_md)

    return demo
````

````python
# ==========================================
# 실습 2: Button 컴포넌트
# ==========================================

def button_demo():
    """Button 컴포넌트 데모"""
    with gr.Blocks(theme=gr.themes.Soft()) as demo:
        gr.Markdown("# Button 컴포넌트 실습")

        with gr.Tab("기본 버튼"):
            gr.Markdown("### 다양한 스타일의 버튼")

            with gr.Row():
                btn1 = gr.Button("기본 버튼")
                btn2 = gr.Button("Primary", variant="primary")
                btn3 = gr.Button("Secondary", variant="secondary")
                btn4 = gr.Button("Stop", variant="stop")

            output1 = gr.Textbox(label="버튼 클릭 결과", interactive=False)

            btn1.click(lambda: "기본 버튼 클릭됨!", outputs=output1)
            btn2.click(lambda: "Primary 버튼 클릭됨!", outputs=output1)
            btn3.click(lambda: "Secondary 버튼 클릭됨!", outputs=output1)
            btn4.click(lambda: "Stop 버튼 클릭됨!", outputs=output1)

        with gr.Tab("버튼 크기"):
            gr.Markdown("### size 파라미터로 크기 조절")

            with gr.Row():
                small_btn = gr.Button("Small", size="sm")
                medium_btn = gr.Button("Medium (기본)")
                large_btn = gr.Button("Large", size="lg")

            output2 = gr.Textbox(label="결과")

            small_btn.click(lambda: "Small 버튼!", outputs=output2)
            medium_btn.click(lambda: "Medium 버튼!", outputs=output2)
            large_btn.click(lambda: "Large 버튼!", outputs=output2)

        with gr.Tab("카운터 예제"):
            gr.Markdown("### 버튼으로 상태 관리하기")

            counter_state = gr.State(0)
            counter_display = gr.Number(label="카운터", value=0, interactive=False)

            with gr.Row():
                minus_btn = gr.Button("감소")
                reset_btn = gr.Button("초기화", variant="secondary")
                plus_btn = gr.Button("증가", variant="primary")

            def increment(count):
                return count + 1, count + 1

            def decrement(count):
                return count - 1, count - 1

            def reset():
                return 0, 0

            plus_btn.click(
                fn=increment,
                inputs=counter_state,
                outputs=[counter_state, counter_display]
            )
            minus_btn.click(
                fn=decrement,
                inputs=counter_state,
                outputs=[counter_state, counter_display]
            )
            reset_btn.click(
                fn=reset,
                outputs=[counter_state, counter_display]
            )

        with gr.Tab("진행률 표시"):
            gr.Markdown("### 버튼으로 긴 작업 실행")

            progress_btn = gr.Button("작업 시작 (5초)", variant="primary")
            progress_output = gr.Textbox(label="진행 상태", lines=5)

            def long_task():
                result = ""
                for i in range(5):
                    time.sleep(1)
                    result += f" {i+1}초 경과...\n"
                    yield result
                yield result + "\n 작업 완료!"

            progress_btn.click(fn=long_task, outputs=progress_output)

    return demo
````

````python
# ==========================================
# 실습 3: Tab 컴포넌트
# ==========================================

def tab_demo():
    """Tab 컴포넌트 데모"""
    with gr.Blocks(theme=gr.themes.Soft()) as demo:
        gr.Markdown("# Tab 컴포넌트 실습")

        with gr.Tab("소개"):
            gr.Markdown("""
            ## Tab 컴포넌트란?

            Tab은 여러 화면을 탭으로 구분하여 표시하는 컴포넌트입니다.

            **장점:**
            - 공간 효율적
            - 직관적인 네비게이션
            - 관련 기능 그룹화
            """)

        with gr.Tab("계산기"):
            gr.Markdown("### 간단한 계산기")

            with gr.Row():
                num1 = gr.Number(label="숫자 1", value=10)
                operation = gr.Radio(
                    label="연산",
                    choices=["더하기", "빼기", "곱하기", "나누기"],
                    value="더하기"
                )
                num2 = gr.Number(label="숫자 2", value=5)

            calc_btn = gr.Button("계산", variant="primary")
            calc_result = gr.Number(label="결과")

            def calculate(n1, op, n2):
                ops = {
                    "더하기": n1 + n2,
                    "빼기": n1 - n2,
                    "곱하기": n1 * n2,
                    "나누기": n1 / n2 if n2 != 0 else "오류: 0으로 나눌 수 없음"
                }
                return ops[op]

            calc_btn.click(
                fn=calculate,
                inputs=[num1, operation, num2],
                outputs=calc_result
            )

        with gr.Tab("텍스트 도구"):
            gr.Markdown("### 텍스트 변환 도구")

            text_input = gr.Textbox(
                label="입력",
                placeholder="텍스트를 입력하세요...",
                lines=3
            )

            with gr.Row():
                upper_btn = gr.Button("대문자")
                lower_btn = gr.Button("소문자")
                reverse_btn = gr.Button("역순")
                count_btn = gr.Button("단어 수", variant="secondary")

            text_output = gr.Textbox(label="결과", lines=3)

            upper_btn.click(lambda x: x.upper(), inputs=text_input, outputs=text_output)
            lower_btn.click(lambda x: x.lower(), inputs=text_input, outputs=text_output)
            reverse_btn.click(lambda x: x[::-1], inputs=text_input, outputs=text_output)
            count_btn.click(
                lambda x: f"단어 수: {len(x.split())}개\n글자 수: {len(x)}개",
                inputs=text_input,
                outputs=text_output
            )

        with gr.Tab("통계"):
            gr.Markdown("### 탭 사용 통계")
            gr.Markdown("""
            이 데모는 4개의 탭으로 구성되어 있습니다:

            1. **소개** - Tab 컴포넌트 설명
            2. **계산기** - 사칙연산 도구
            3. **텍스트 도구** - 텍스트 변환
            4. **통계** - 현재 페이지

            > **팁**: 탭을 클릭하여 각 기능을 확인해보세요!
            """)

    return demo
````

````python
# ==========================================
# 실습 4: Row & Column 레이아웃
# ==========================================

def layout_demo():
    """Row와 Column 레이아웃 데모"""
    with gr.Blocks(theme=gr.themes.Soft()) as demo:
        gr.Markdown("# Row & Column 레이아웃 실습")

        with gr.Tab("기본 레이아웃"):
            gr.Markdown("### Row - 가로 배치")

            with gr.Row():
                gr.Textbox(label="입력 1", placeholder="첫 번째")
                gr.Textbox(label="입력 2", placeholder="두 번째")
                gr.Textbox(label="입력 3", placeholder="세 번째")

            gr.Markdown("### Column - 세로 배치")

            with gr.Column():
                gr.Textbox(label="입력 A", placeholder="위")
                gr.Textbox(label="입력 B", placeholder="중간")
                gr.Textbox(label="입력 C", placeholder="아래")

        with gr.Tab("비율 조정 (scale)"):
            gr.Markdown("### scale 파라미터로 너비 비율 조정")

            gr.Markdown("#### 1:1:1 비율 (기본)")
            with gr.Row():
                gr.Textbox(label="1")
                gr.Textbox(label="1")
                gr.Textbox(label="1")

            gr.Markdown("#### 2:1:1 비율")
            with gr.Row():
                gr.Textbox(label="2", scale=2)
                gr.Textbox(label="1", scale=1)
                gr.Textbox(label="1", scale=1)

            gr.Markdown("#### 3:1 비율")
            with gr.Row():
                gr.Textbox(label="3", scale=3)
                gr.Textbox(label="1", scale=1)

        with gr.Tab("중첩 레이아웃"):
            gr.Markdown("### Row와 Column을 중첩하여 복잡한 레이아웃 구성")

            with gr.Row():
                with gr.Column(scale=2):
                    gr.Markdown("#### 왼쪽 영역 (넓음)")
                    input1 = gr.Textbox(label="제목")
                    input2 = gr.Textbox(label="내용", lines=5)
                    submit_btn = gr.Button("제출", variant="primary")

                with gr.Column(scale=1):
                    gr.Markdown("#### 오른쪽 영역 (좁음)")
                    gr.Markdown("**옵션**")
                    option1 = gr.Checkbox(label="공개")
                    option2 = gr.Checkbox(label="알림")
                    category = gr.Dropdown(
                        label="카테고리",
                        choices=["공지", "질문", "자유"]
                    )

        with gr.Tab("실전 예제"):
            gr.Markdown("### 프로필 편집기")

            with gr.Row():
                # 왼쪽: 입력 폼
                with gr.Column(scale=2):
                    gr.Markdown("#### 기본 정보")
                    profile_name = gr.Textbox(label="이름", placeholder="홍길동")
                    profile_email = gr.Textbox(label="이메일", placeholder="hong@example.com")

                    with gr.Row():
                        profile_age = gr.Number(label="나이", value=25, minimum=0)
                        profile_gender = gr.Radio(
                            label="성별",
                            choices=["남성", "여성", "기타"]
                        )

                    profile_bio = gr.Textbox(
                        label="자기소개",
                        lines=4,
                        placeholder="자기소개를 입력하세요..."
                    )

                    with gr.Row():
                        clear_btn = gr.Button("초기화", variant="secondary")
                        save_btn = gr.Button("저장", variant="primary")

                # 오른쪽: 미리보기
                with gr.Column(scale=1):
                    gr.Markdown("#### 미리보기")
                    preview = gr.Markdown("프로필이 여기에 표시됩니다.")

            def update_preview(name, email, age, gender, bio):
                return f"""
                ### {name or '이름 없음'}

                **이메일**: {email or 'N/A'}
                **나이**: {age}세
                **성별**: {gender or 'N/A'}

                ---

                {bio or '*자기소개가 없습니다.*'}
                """

            def clear_form():
                return "", "", 25, None, "", "프로필이 초기화되었습니다."

            # 실시간 미리보기
            for component in [profile_name, profile_email, profile_age, profile_gender, profile_bio]:
                component.change(
                    fn=update_preview,
                    inputs=[profile_name, profile_email, profile_age, profile_gender, profile_bio],
                    outputs=preview
                )

            save_btn.click(
                fn=lambda n, e, a, g, b: f"저장 완료!\n\n{update_preview(n, e, a, g, b)}",
                inputs=[profile_name, profile_email, profile_age, profile_gender, profile_bio],
                outputs=preview
            )

            clear_btn.click(
                fn=clear_form,
                outputs=[profile_name, profile_email, profile_age, profile_gender, profile_bio, preview]
            )

    return demo
````

````python
# ==========================================
# 통합 데모 실행
# ==========================================

# 모든 데모를 하나의 앱으로 통합
with gr.Blocks(theme=gr.themes.Soft(), title="Gradio 기본 컴포넌트") as demo:
    gr.Markdown("# Gradio 기본 컴포넌트 종합 실습")
    gr.Markdown("각 탭에서 다양한 컴포넌트를 실습할 수 있습니다.")

    with gr.Tab("Markdown"):
        markdown_demo()

    with gr.Tab("Button"):
        button_demo()

    with gr.Tab("Tab"):
        tab_demo()

    with gr.Tab("Layout"):
        layout_demo()

demo.launch()
````
