#Hello World - 가장 기본적인 Gradio 앱

```python
import gradio as gr

def greet(name):
    """간단한 인사 함수"""
    if not name:
        return "이름을 입력해주세요!"
    return f"안녕하세요, {name}님!"
```

```python
# Interface 생성
demo = gr.Interface(
    fn=greet,
    inputs=gr.Textbox(label="이름", placeholder="이름을 입력하세요..."),
    outputs=gr.Textbox(label="인사"),
    title="Hello World",
    description="Gradio의 가장 기본적인 예제입니다.",
    examples=[["철수"], ["영희"], ["Gradio"]]
)
```

```python
demo.launch()
```
